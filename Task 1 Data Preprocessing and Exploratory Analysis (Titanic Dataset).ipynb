Task 1: Data Preprocessing and Exploratory Analysis (Titanic Dataset)
● Exam Task:
1. Load the Titanic dataset using Pandas and handle missing values by imputing
missing Age values with the median
2. Filter the dataset to extract all surviving passengers (Survived == 1) and
compute the average and maximum Fare using NumPy operations
3. Using Seaborn or Matplotlib, create two visualizations: a survival count bar plot
and an Age vs. Fare scatter plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the Titanic dataset
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Impute missing Age values with the median
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)

print(f"Missing Age values imputed with median: {median_age}")
display(df.head())
```

```python
# 2. Filter surviving passengers and compute Fare stats using NumPy
surviving_passengers = df[df['Survived'] == 1]
fares = surviving_passengers['Fare'].to_numpy()

average_fare = np.mean(fares)
max_fare = np.max(fares)

print(f"Surviving Passengers Count: {len(surviving_passengers)}")
print(f"Average Fare of Survivors: ${average_fare:.2f}")
print(f"Maximum Fare of Survivors: ${max_fare:.2f}")
```

```python
# 3. Create visualizations
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Survival count bar plot
sns.countplot(ax=axes[0], data=df, x='Survived', palette='Set2')
axes[0].set_title('Survival Count (0 = No, 1 = Yes)')
axes[0].set_xlabel('Survived')
axes[0].set_ylabel('Count')

# Age vs. Fare scatter plot
sns.scatterplot(ax=axes[1], data=df, x='Age', y='Fare', hue='Survived', palette='coolwarm', alpha=0.7)
axes[1].set_title('Passenger Age vs. Fare')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Fare')

plt.tight_layout()
plt.show()
```
